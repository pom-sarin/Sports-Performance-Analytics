```python
import pandas as pd
import numpy as np
```

```python
NBAmod = pd.read_excel("Assignment Data/Week 4/NBA prediction model (Assignment).xlsx")
NBAmod
```

```python
NBAmod['winprob']= 1/(NBAmod['hwinodds'])/(1/(NBAmod['hwinodds'])+ 1/(NBAmod['hloseodds']))
NBAmod['winpred'] = np.where(NBAmod['winprob'] >= .5,1,0)
NBAmod
```

```python
# Taking logs of team and opponent salaries
NBAmod['lg_team_sal']=np.log(NBAmod['hteamsal'])
NBAmod['lg_opp_sal']=np.log(NBAmod['opposal'])

# taking salary ratio between two teams
NBAmod['lg_sal_ratio']= NBAmod['lg_team_sal'] - NBAmod['lg_opp_sal']
NBAmod
```

```python
NBA18 = NBAmod[NBAmod['year']==2018]
from sklearn.linear_model import LogisticRegression
import statsmodels.api as sm
import statsmodels.formula.api as smf
Win_lgsal = 'hwin~lg_sal_ratio'
logit_model = smf.glm(formula = Win_lgsal, data = NBA18, family=sm.families.Binomial())
result = logit_model.fit()
print(result.summary())
```

```python
NBAmod['predW'] = 1/(1+np.exp(-(0.4452+1.1216*NBAmod['lg_sal_ratio'])))
NBAmod['predL'] = 1- NBAmod['predW'] 
display(NBAmod[0:10])
```

```python
NBAmod.loc[NBAmod.predL > NBAmod.predW, 'fitted'] = 0
NBAmod.loc[NBAmod.predW > NBAmod.predL, 'fitted'] = 1
NBAmod
```

```python
NBA19 = NBAmod[NBAmod['year']==2019]
NBA19['TRUE_odds'] = np.where(NBA19['winpred']==NBA19['hwin'], 1, 0)
NBA19['TRUE_logit'] = np.where(NBA19['fitted']==NBA19['hwin'], 1, 0)
display(NBA19[0:10])
```
