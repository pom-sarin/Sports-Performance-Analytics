# I. Data Coding and Merging

```python
# Load packages
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Load game logs data
Teams = pd.read_excel("../Data/Game logs 1999-2003.xlsx")
```

```python
#Your Code Here
Teams['hwin'] = np.where(Teams['home_score'] > Teams['visitor_score'],1,0)
Teams['awin'] = np.where(Teams['home_score'] < Teams['visitor_score'],1,0)
```

```python
Teams['year'] = Teams['date'].astype(str).str[0:4]
Teams
```

```python
Teamshome = Teams.groupby(['home','year'])['visitor_ab', 'visitor_h','home_ab', 'home_h', 'hwin','awin'].sum()
Teamshome = Teamshome.reset_index()
Teamshome = Teamshome.rename(columns = {"home": "Team"}) 
Teamshome
```

```python
Teamsaway = Teams.groupby(['visitor','year'])['visitor_ab', 'visitor_h','home_ab', 'home_h', 'hwin','awin'].sum()
Teamsaway = Teamsaway.reset_index()
Teamsaway = Teamsaway.rename(columns = {"visitor": "Team"}) 
Teamsaway
```

```python
Teams2 = pd.merge(Teamshome, Teamsaway, on =['Team','year'])
Teams2['wins'] = Teams2['hwin_x']+Teams2['awin_y']
Teams2
```

```python
season_win = Teams2.groupby('year')['home_ab_y'].mean()
```

```python
season_win
```

# II. Calculating Performance Metrics

```python
#Your Code Here
Teams2['BAFOR'] = (Teams2['home_h_x']+Teams2['visitor_h_y'])/(Teams2['home_ab_x']+Teams2['visitor_ab_y'])
Teams2['BAAGN'] = (Teams2['home_h_y']+Teams2['visitor_h_x'])/(Teams2['home_ab_y']+Teams2['visitor_ab_x'])
```

```python
Teams3 = Teams2[['year', 'Team', 'wins','BAFOR','BAAGN']]
Teams3
```

```python
TeamsGh = Teams.groupby(['year','home'])['hwin'].count()
TeamsGh = TeamsGh.reset_index()
TeamsGh = TeamsGh.rename(columns = {"home": 'Team'}) 

TeamsGa = Teams.groupby(['year','visitor'])['awin'].count()
TeamsGa = TeamsGa.reset_index()
TeamsGa = TeamsGa.rename(columns = {"visitor": 'Team'}) 

TeamsG = pd.merge(TeamsGh, TeamsGa, on =['Team','year'])
TeamsG['Games'] = TeamsG['hwin']+TeamsG['awin']
TeamsG = TeamsG[['year','Team','Games']]
TeamsG
```

```python
Teams3 = pd.merge(Teams3, TeamsG, on =['Team','year'])
Teams3
```

```python
Teams3['wpc'] = Teams3['wins']/Teams3['Games']
Teams3
```

```python
Teams3['BA_DIFF'] = Teams3['BAFOR'] - Teams3['BAAGN']
Teams3.sort_values(by='BA_DIFF')
```

```python
Teams3[Teams3['Games'] != 162].shape
```

```python
Teamswpc = Teams2[['year','Team','hwin_x','awin_y']]
Teamswpc
```

```python
TeamsGh = TeamsGh.rename(columns = {"hwin": 'hgame'})
TeamsGa = TeamsGa.rename(columns = {"awin": 'agame'})
Teamswpc = pd.merge(Teamswpc, TeamsGh, on =['Team','year'])
Teamswpc = pd.merge(Teamswpc, TeamsGa, on =['Team','year'])
Teamswpc
```

```python
Teamswpc['wpc_h'] = Teamswpc['hwin_x']/Teamswpc['hgame']
Teamswpc['wpc_a'] = Teamswpc['awin_y']/Teamswpc['agame']
```

```python
Teamswpc['wpc_diff'] = Teamswpc['wpc_h'] - Teamswpc['wpc_a']
Teamswpc
```

```python
Teamswpc.sort_values(by='wpc_diff',ascending=False)
```

# III. Running Regressions

```python
#Your Code Here
import statsmodels.formula.api as smf
WinBA_lm = smf.ols(formula = 'wpc ~ BAFOR + BAAGN', data=Teams3).fit()
WinBA_lm.summary()
```

```python
WinBAR_lm = smf.ols(formula = 'wpc ~ I(BAFOR - BAAGN)', data=Teams3).fit()
WinBAR_lm.summary()
```

```python
from statsmodels.iolib.summary2 import summary_col
Header = ['1','2']
Table_1 = summary_col([WinBA_lm,WinBAR_lm],\
                      regressor_order=['Intercept','BAFOR','BAAGN'],model_names = Header)
print(Table_1)
```
